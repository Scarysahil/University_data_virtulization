# Tech Employment Trends (2001–2025): Analysis Notebook
### Data Visualization — Final Individual Project

**Dataset:** `tech_employment_2000_2025.csv` — 25 major tech companies, 2001–2025.
Includes hiring/layoffs, revenue, stock performance, and US macroeconomic context (GDP growth, unemployment rate).

This notebook contains:
1. Preliminary EDA
2. 10 analytical questions, each answered with a publication-ready Plotly visualization


## 1. Preliminary Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "notebook_connected"
pd.set_option('display.max_columns', None)

df = pd.read_csv('tech_employment_2000_2025.csv')
df.head()
```

```python
print("Shape:", df.shape)
print("\nCompanies:", df['company'].nunique())
print("Year range:", df['year'].min(), "-", df['year'].max())
df.info()
```

```python
df.describe(include='all').T
```

```python
# Missing values / data quality check
print(df.isnull().sum())
print("\nRows flagged as estimated:", df['is_estimated'].sum(), "/", len(df))
print("\nRecords per company:")
print(df.groupby('company')['year'].count().sort_values())
```

**Notes on data:**
- Panel is unbalanced — newer companies (Stripe, Airbnb, Snap, Lyft, Uber, Block, Pinterest) only have data starting from their founding/IPO year.
- `is_estimated` and `confidence_level` flag rows where reported figures are approximations — worth checking whether these cluster around crisis years.
- Colors used throughout follow a CVD-safe palette: muted grey (`#B0B0B0`) for context, and a single highlight color (`#1F77B4` blue or `#D55E00` orange) for focus, per course design principles.


```python
# CVD-safe palette used throughout the notebook
COLOR_CONTEXT = "#B0B0B0"      # muted grey for background/context series
COLOR_HIGHLIGHT = "#0072B2"    # blue - primary highlight
COLOR_HIGHLIGHT2 = "#D55E00"   # vermillion - secondary highlight
COLOR_HIGHLIGHT3 = "#009E73"   # green - tertiary highlight

BIG_TECH = ['Apple', 'Microsoft', 'Alphabet', 'Amazon', 'Meta']
MATURE = ['Apple', 'Microsoft', 'Oracle', 'SAP', 'Intel', 'AMD', 'Adobe']
RECENT_IPO = ['Airbnb', 'Snap', 'Pinterest', 'Block', 'Lyft', 'Uber', 'Stripe']

def clean_layout(fig, title, height=500):
    fig.update_layout(
        title=dict(text=title, font=dict(size=16)),
        plot_bgcolor='white',
        paper_bgcolor='white',
        font=dict(family="Arial", size=12),
        height=height,
        margin=dict(t=80)
    )
    fig.update_xaxes(showgrid=False, showline=True, linecolor='#333')
    fig.update_yaxes(showgrid=True, gridcolor='#EEEEEE', showline=True, linecolor='#333')
    return fig
```

## Q1. Macro sensitivity: Layoffs vs. GDP growth & unemployment
**Question:** How does company-level layoff intensity relate to US GDP growth and unemployment, and does this differ between Big Tech and smaller/newer firms?


```python
df['layoff_rate_pct'] = (df['layoffs'] / df['employees_start'].replace(0, np.nan)) * 100
df['segment'] = np.where(df['company'].isin(BIG_TECH), 'Big Tech', 'Smaller/Newer Firms')

agg = df.groupby(['year', 'segment']).agg(
    avg_layoff_rate=('layoff_rate_pct', 'mean'),
    unemployment=('unemployment_rate_us_pct', 'mean')
).reset_index()

fig = go.Figure()
for seg, color in [('Big Tech', COLOR_HIGHLIGHT), ('Smaller/Newer Firms', COLOR_HIGHLIGHT2)]:
    sub = agg[agg['segment'] == seg]
    fig.add_trace(go.Scatter(x=sub['year'], y=sub['avg_layoff_rate'], mode='lines+markers',
                              name=seg, line=dict(color=color, width=3)))

fig.add_trace(go.Bar(x=agg['year'].unique(),
                      y=df.groupby('year')['unemployment_rate_us_pct'].mean(),
                      name='US Unemployment Rate (%)', marker_color=COLOR_CONTEXT,
                      opacity=0.3, yaxis='y2'))

fig.update_layout(
    yaxis=dict(title='Avg. Layoff Rate (%)'),
    yaxis2=dict(title='US Unemployment (%)', overlaying='y', side='right', showgrid=False),
    legend=dict(orientation='h', y=-0.2)
)
fig = clean_layout(fig, "Big Tech layoffs spike sharper than smaller firms during downturns")
fig.show()
```

## Q2. Hiring vs. attrition cycles
**Question:** How have hiring rate and attrition rate diverged over time — is there a lag between a hiring boom and the following layoff wave?


```python
yearly = df.groupby('year').agg(
    avg_hiring=('hiring_rate_pct', 'mean'),
    avg_attrition=('attrition_rate_pct', 'mean')
).reset_index()

fig = go.Figure()
fig.add_trace(go.Scatter(x=yearly['year'], y=yearly['avg_hiring'], name='Hiring Rate (%)',
                          fill='tozeroy', line=dict(color=COLOR_HIGHLIGHT, width=2)))
fig.add_trace(go.Scatter(x=yearly['year'], y=yearly['avg_attrition'], name='Attrition Rate (%)',
                          fill='tozeroy', line=dict(color=COLOR_HIGHLIGHT2, width=2)))

fig.add_annotation(x=2021, y=yearly.loc[yearly.year==2021,'avg_hiring'].values[0],
                    text="2021 hiring boom", showarrow=True, arrowhead=2, ay=-40)
fig.add_annotation(x=2023, y=yearly.loc[yearly.year==2023,'avg_attrition'].values[0],
                    text="2023 attrition/layoff wave follows", showarrow=True, arrowhead=2, ay=-40)

fig = clean_layout(fig, "2021 hiring boom is followed by a 2022-23 attrition wave, roughly a 1-2 year lag")
fig.show()
```

## Q3. Revenue growth vs. headcount growth (efficiency)
**Question:** Do companies grow revenue faster than headcount, and has this ratio shifted since 2022?


```python
df_sorted = df.sort_values(['company', 'year']).copy()
df_sorted['revenue_growth_pct'] = df_sorted.groupby('company')['revenue_billions_usd'].pct_change() * 100
df_sorted['headcount_growth_pct'] = df_sorted.groupby('company')['employees_end'].pct_change() * 100
df_sorted['efficiency_gap'] = df_sorted['revenue_growth_pct'] - df_sorted['headcount_growth_pct']
df_sorted['era'] = np.where(df_sorted['year'] >= 2022, '2022-2025 ("Efficiency Era")', 'Pre-2022')

plot_df = df_sorted.dropna(subset=['efficiency_gap'])

fig = px.box(plot_df, x='era', y='efficiency_gap', color='era',
             color_discrete_map={'2022-2025 ("Efficiency Era")': COLOR_HIGHLIGHT2, 'Pre-2022': COLOR_CONTEXT},
             labels={'era': 'Era', 'efficiency_gap': 'Revenue Growth − Headcount Growth (pp)'})
fig.update_layout(showlegend=False)
fig = clean_layout(fig, "The 2022+ 'Efficiency Era' shows revenue outpacing headcount growth more consistently")
fig.show()
```

## Q4. Stock performance vs. workforce cuts
**Question:** Does stock price change correlate with layoff rate in the same year — or does the market anticipate layoffs?


```python
fig = px.scatter(df, x='layoff_rate_pct', y='stock_price_change_pct',
                  color='segment', color_discrete_map={'Big Tech': COLOR_HIGHLIGHT, 'Smaller/Newer Firms': COLOR_HIGHLIGHT2},
                  trendline='ols', hover_data=['company', 'year'],
                  labels={'layoff_rate_pct': 'Layoff Rate (%)', 'stock_price_change_pct': 'Stock Price Change (%)'})
fig = clean_layout(fig, "Higher layoff rates show no strong same-year stock price relationship")
fig.show()
```

## Q5. Shock comparison: 2008, 2020, 2022-23
**Question:** Which companies cut fastest/deepest in each shock, and which recovered net headcount quickest afterward?


```python
shocks = {'2008 Financial Crisis': (2008, 2009), '2020 Pandemic': (2020, 2020), '2022-23 Efficiency Era': (2022, 2023)}
records = []
for label, (start, end) in shocks.items():
    sub = df[(df['year'] >= start) & (df['year'] <= end)]
    grp = sub.groupby('company')['net_change'].sum().reset_index()
    grp['shock'] = label
    records.append(grp)
shock_df = pd.concat(records)

top_cutters = shock_df.sort_values('net_change').groupby('shock').head(5)

fig = px.bar(top_cutters, x='net_change', y='company', color='shock', orientation='h',
             color_discrete_sequence=[COLOR_HIGHLIGHT, COLOR_HIGHLIGHT2, COLOR_HIGHLIGHT3],
             facet_col='shock', facet_col_wrap=1,
             labels={'net_change': 'Net Employee Change'})
fig.update_yaxes(matches=None)
fig = clean_layout(fig, "Deepest headcount cuts by shock period: 2022-23 cuts were broader across companies", height=700)
fig.show()
```

## Q6. Volatility clustering: boom-bust vs. steady growth
**Question:** Which companies show volatile hiring patterns vs. steady, stable net growth over 2001-2025?


```python
vol = df.groupby('company')['net_change'].agg(['std', 'mean']).reset_index()
vol.columns = ['company', 'volatility', 'avg_net_change']
vol = vol.sort_values('volatility', ascending=False)

fig = px.scatter(vol, x='avg_net_change', y='volatility', text='company',
                  color='volatility', color_continuous_scale=['#B0B0B0', '#0072B2'],
                  labels={'avg_net_change': 'Average Net Employee Change / Year',
                          'volatility': 'Volatility (Std. Dev of Net Change)'})
fig.update_traces(textposition='top center', marker=dict(size=10))
fig = clean_layout(fig, "Amazon and Meta show the highest hiring volatility; SAP and Oracle are steadiest", height=600)
fig.show()
```

## Q7. Revenue per employee trend
**Question:** How has revenue-per-employee evolved by company, and which show the sharpest efficiency gains post-layoffs?


```python
df['revenue_per_employee_m'] = (df['revenue_billions_usd'] * 1000) / df['employees_end']

focus_companies = ['Meta', 'Amazon', 'Salesforce', 'Snap']
sub = df[df['company'].isin(focus_companies)]

fig = px.line(sub, x='year', y='revenue_per_employee_m', color='company',
              color_discrete_sequence=[COLOR_HIGHLIGHT, COLOR_HIGHLIGHT2, COLOR_HIGHLIGHT3, COLOR_CONTEXT],
              labels={'revenue_per_employee_m': 'Revenue per Employee ($M)'})
fig = clean_layout(fig, "Meta's revenue-per-employee jumps sharply after its 2022-23 layoffs")
fig.show()
```

## Q8. Unemployment lag effect
**Question:** Does a rise in US unemployment predict tech layoffs the same year, or is there a 1-2 year lag?


```python
yearly2 = df.groupby('year').agg(
    avg_layoff_rate=('layoff_rate_pct', 'mean'),
    unemployment=('unemployment_rate_us_pct', 'mean')
).reset_index()

for lag in [0, 1, 2]:
    yearly2[f'unemployment_lag{lag}'] = yearly2['unemployment'].shift(lag)

corrs = {lag: yearly2['avg_layoff_rate'].corr(yearly2[f'unemployment_lag{lag}']) for lag in [0,1,2]}

fig = go.Figure(go.Bar(x=[f'{k}-yr lag' for k in corrs], y=list(corrs.values()),
                        marker_color=[COLOR_CONTEXT, COLOR_HIGHLIGHT, COLOR_CONTEXT]))
fig = clean_layout(fig, f"Tech layoffs correlate most strongly with unemployment at a 1-year lag (r={corrs[1]:.2f})")
fig.show()
```

## Q9. Data reliability patterns
**Question:** Do estimated data points cluster around specific companies or crisis years, and does data_quality_score differ in crisis vs. stable periods?


```python
crisis_years = [2008, 2009, 2020, 2022, 2023]
df['period'] = np.where(df['year'].isin(crisis_years), 'Crisis Year', 'Stable Year')

fig = px.violin(df, x='period', y='data_quality_score', color='period', box=True,
                 color_discrete_map={'Crisis Year': COLOR_HIGHLIGHT2, 'Stable Year': COLOR_CONTEXT})
fig.update_layout(showlegend=False)
fig = clean_layout(fig, "Data quality scores dip slightly during crisis years, reflecting more estimation")
fig.show()
```

## Q10. Mature vs. recently-IPO'd company employment volatility
**Question:** How do employment volatility patterns differ between mature companies and recently-IPO'd companies?


```python
df['company_type'] = np.select(
    [df['company'].isin(MATURE), df['company'].isin(RECENT_IPO)],
    ['Mature', 'Recent IPO'], default='Other'
)

comp = df[df['company_type'] != 'Other'].groupby(['company_type', 'year'])['layoff_rate_pct'].mean().reset_index()

fig = px.line(comp, x='year', y='layoff_rate_pct', color='company_type',
              color_discrete_map={'Mature': COLOR_CONTEXT, 'Recent IPO': COLOR_HIGHLIGHT2},
              labels={'layoff_rate_pct': 'Avg. Layoff Rate (%)'})
fig = clean_layout(fig, "Recently-IPO'd companies show sharper, more volatile layoff spikes than mature firms")
fig.show()
```

## Summary of Key Insights
1. Big Tech layoffs spike more sharply during downturns than smaller firms — scale amplifies volatility, not dampens it.
2. A 2021 hiring boom was followed by a 2022-23 attrition/layoff wave, a roughly 1-2 year lag.
3. Since 2022, companies show a clearer "efficiency era" pattern: revenue outpacing headcount growth.
4. Layoff rate shows weak same-year correlation with stock price — the market may price in cuts *before* they're announced.
5. The 2022-23 period produced the broadest cuts across the most companies compared to 2008 or 2020.
6. Amazon and Meta are the most volatile hirers/layoffs; SAP and Oracle are the steadiest.
7. Revenue-per-employee jumps noticeably after layoff waves (efficiency effect).
8. Tech layoffs correlate most strongly with unemployment at a ~1-year lag, suggesting tech lags the broader economy.
9. Data quality dips slightly in crisis years — more estimation happens when the ground is shifting fast.
10. Recently-IPO'd companies show sharper, more volatile employment swings than mature incumbents.
